```python
def drop_colums(df,colms):
    '''
    Burada daha basit kod görüntüsü oluşturmak için fonksiyon oluşturuldu.
    '''
    for col in colms:
        try:
            df.drop([col], axis=1, inplace=True)
        except Exception as e:
            print(f"Bir hata oluştu: {e}")
```

```python
# şimdi ise datamız üzerinde nasıl yapabiliriz ona bakalım: 
df['location'] = df['img-link href'].apply(lambda x: x.split('/')[3])  #lambda anonim fonksiyondur.
df['city'] = df['location'].str.split('-').str[0]
df['district'] = df['location'].str.split('-').str[1]
df['neighborhood'] = df['location'].str.split('-').str[2:-1]
df['neighborhood'] = df['neighborhood'].apply(lambda x: '_'.join(x))
print(df['neighborhood'].unique())
```
